Multimodal AI is a field of AI that deals with the interaction of multiple modalities (e.g. text, image, audio, video, etc.)

- 💬 **Textual data:** Natural language, documents

## Why Multimodal Is Central to Agentic AI

Multimodal capability turns LLMs from chat systems into **grounded agents** that can perceive, reason, and act.

- **Perceive real environments**
  - Read screens, documents, charts, photos, and video frames instead of text-only descriptions.

- **Ground decisions in evidence**
  - Use visual + language context to choose actions based on what is actually present.

- **Act through tools and interfaces**
  - Navigate apps/web UIs, call APIs, and execute workflows from multimodal observations.

- **Operate in feedback loops**
  - Plan -> observe -> act -> verify over many steps for long-horizon tasks.

- **Collaborate naturally with humans**
  - Accept instructions and corrections via text, voice, and visual references.

**Takeaway:** Classic tasks like captioning and VQA are useful benchmarks, but the high-impact application is building reliable multimodal agents.

## Vision-Language Models

Vision-Language models combine computer vision and natural language processing.

**Key Components:**

1. **Image Encoder**: Typically a CNN or Vision Transformer
2. **Text Encoder**: Typically a transformer-based language model
3. **Cross-Modal Fusion**: Mechanisms to align and combine visual and textual features
4. **Task-Specific Head**: Output layer for the specific task (classification, generation, etc.)

**Popular Vision-Language Models:**

- [CLIP](https://arxiv.org/pdf/2103.00020): Contrastive Language-Image Pre-training 
- [DALL-E](https://arxiv.org/pdf/2102.12092): Text-to-image generation
- [Flamingo](https://arxiv.org/pdf/2204.14198): Few-shot learning across vision and language tasks
- [PaLI](https://arxiv.org/pdf/2209.06794): Scaling multilingual language-image learning

*Large scale models trained on massive datasets are often refereed to as Foundation Models*


## Model Taxonomy at a Glance

| Family | Typical Input -> Output | Training Signal | Fusion Style | Example | Best For |
|---|---|---|---|---|---|
| Contrastive Alignment | image + text -> embeddings | contrastive (match/non-match) | late or weak hybrid | CLIP | retrieval, zero-shot classification |
| Discriminative Multimodal | image + text -> label/span/answer | supervised task loss | early/hybrid | LayoutLMv3, VQA models | classification, structured prediction |
| Generative VLM (text output) | image + text -> text | autoregressive LM | hybrid cross-attention | Flamingo, PaLI | captioning, VQA, dialog |
| Generative T2I | text -> image | diffusion/AR image loss | n/a (text-conditioned generator) | DALL-E, SDXL | image synthesis |

## Representation Alignment {.scrollable}

How the modalities connect.

This is the core technical problem: aligning different signal spaces.

**Contrastive Alignment**

- Learn a shared embedding space
- Pull matched pairs together, push mismatched apart
- Example: CLIP

Key idea:
“Do these belong together?”

**Generative Modeling**

- Model joint distribution across modalities
- Can generate one modality from another
- Examples:
  - Image → text (captioning)
  - Text → image (diffusion models)
  - Multimodal LLMs like GPT-4V

Key idea:
“Given X, generate Y”

**Token Unification**

- Convert all modalities into a common token space
- Feed into a single transformer
- Example: PaLI

Key idea:
"All inputs are just tokens"

## Fusion Strategies {.scrollable}

Where modalities interact

**Early Fusion**

- Combine raw inputs early
- Example: concatenating embeddings

Pros: strong interaction
Cons: hard to scale across modalities

**Late Fusion**

- Process each modality independently
- Combine only at decision layer

Pros: modular
Cons: weak cross-modal reasoning

**Cross-Attention (dominant paradigm)**

- One modality attends to another
- Example: Flamingo

Key idea:
“Language queries the image”

**Unified Transformer**

- All tokens processed jointly
- No explicit separation

Trend: increasingly popular in large-scale systems

## Training Paradigms {.scrollable}

How modern multimodal systems are trained.

**Paired Supervision**

- Image-text pairs (alt-text, captions)
- Foundation of CLIP-style alignment

**Web-Scale Weak Supervision**

- Noisy internet data with broad coverage
- Scale improves transfer despite label noise

**Instruction Tuning**

- Aligns models to task intent and user preferences
- Enables assistant-style multimodal behavior

**In-Context Learning**

- Few-shot adaptation at inference time
- Example: Flamingo-style prompted reasoning

## Evaluation {.scrollable}

How we determine whether a system is useful and reliable.

**Task Performance**

- Task-specific metrics (retrieval, VQA, captioning, generation)
- Report metrics by task, not as a single score

**Grounding and Faithfulness**

- Verify outputs are supported by visual/context evidence
- Detect hallucinations and modality neglect

**Robustness**

- Missing/noisy modality stress tests
- Shift, ambiguity, and adversarial prompt evaluations

**Human-Centered Assessment**

- Usefulness, factuality, and instruction-following
- Essential for open-ended agentic workflows

- **Data quality over raw scale**: grounded, diverse, and instruction-relevant examples
- **Latency/compute trade-offs**: balance model size, context length, and response speed
- **Safety and monitoring**: detect hallucinations, failures, and unsafe actions in loops
- **Tool reliability**: evaluate full observe -> reason -> act -> verify pipelines

# CLIP (Contrastive Language-Image Pre-training)

## Introduction

- [CLIP]("https://arxiv.org/pdf/2103.00020") was introduced by OpenAI in 2021
- A foundational multimodal model that learns visual concepts from natural language supervision. 
- It enables tasks like zero-shot image classification, image search, and more, without needing task-specific finetuning.

---



## Motivation

Traditional vision models are often trained with manually labeled datasets and specific tasks (e.g., ImageNet classification). CLIP challenges this paradigm by:

- Learning directly from raw text-image pairs.
- Scaling to large internet-scale data.
- Enabling open-vocabulary classification and generalization.



## Overview of the CLIP Architecture

CLIP consists of two main components:

- **Image Encoder**: Typically a ViT (Vision Transformer) or ResNet.
- **Text Encoder**: A Transformer-based model similar to GPT/BERT.

These encoders are trained to map images and their paired captions to a **shared latent embedding space**.

### Key Concept:

- **Contrastive Loss** is used to bring matching image-text pairs closer in the embedding space, and push non-matching pairs apart.



## CLIP Diagram

[![](./img/clip.png)](https://arxiv.org/pdf/2103.00020#page=2)

---

## Training Procedure

### (1) Contrastive Pretraining

- Each image-caption pair is encoded separately into vectors $I_i$ and $T_i$.
- A batch of N image-text pairs produces N image embeddings and N text embeddings.
- The similarity scores are computed via dot product.
- Cross-entropy loss is used over the similarity matrix, encouraging $I_i \cdot T_i$ to be larger than all $I_i \cdot T_j$ for $j \ne i$.

### (2) Embedding Normalization

- Embeddings are L2 normalized to lie on the unit hypersphere, ensuring cosine similarity.

---

## Loss Function {.scrollable}

Let:

- $N$ = batch size  
- $I_i$ = image embedding for the $i$-th image  
- $T_i$ = text embedding for the $i$-th caption  
- $\text{sim}(I, T) = \frac{I \cdot T}{\|I\|\|T\|}$ = cosine similarity between embeddings  
- $\tau$ = learned temperature parameter (scalar)

Then, the **CLIP loss** is the average of two cross-entropy losses:

$$
\mathcal{L}_{\text{CLIP}} = \frac{1}{2N} \sum_{i=1}^N \left[
    -\log \frac{\exp(\text{sim}(I_i, T_i)/\tau)}{\sum_{j=1}^N \exp(\text{sim}(I_i, T_j)/\tau)}
    -\log \frac{\exp(\text{sim}(T_i, I_i)/\tau)}{\sum_{j=1}^N \exp(\text{sim}(T_i, I_j)/\tau)}
\right]
$$

### 🔍 What's Happening?

- The **first term** encourages each image to match its correct caption over all other captions.
- The **second term** does the same for captions matching their correct image.
- The **temperature** $\tau$ controls the sharpness of the distribution; it’s learned during training.

---

## Zero-Shot Prediction with CLIP

:::: {.columns}
::: {.column width="50%"}

After pretraining:

1. Define a set of class prompts (e.g., "a photo of a {object}").
2. Encode each prompt to get text embeddings.
3. Encode the input image.
4. Compute cosine similarities and pick the closest match.

This allows classification on **unseen datasets** without retraining.

:::
::: {.column width="50%"}

[![](./img/clip-classifier.png)](https://arxiv.org/pdf/2103.00020#page=2)

:::
::::

## Encoders in Detail

### Image Encoder:

- ResNet or Vision Transformer (ViT).
- Outputs a single vector representing the whole image.

### Text Encoder:

- Transformer (like GPT or BERT).
- Uses a special token (e.g., [EOS]) to summarize the entire text.
- Only the final token's embedding is used as the text representation.



## Understanding the Latent Space

CLIP learns a **joint latent space** where:

- Semantically similar images and text lie near each other.
- This space can be used for retrieval, clustering, or manipulation.

The model is trained to create a **modality-invariant space** with strong semantic structure.



## Comparison with Other Multimodal Models

| Model     | Goal                         | Input Alignment   | Output |
|-----------|------------------------------|-------------------|--------|
| CLIP      | Image-text retrieval, zero-shot | Separate encoders | Embeddings |
| Flamingo | Vision-language generation   | Interleaved inputs | Text    |
| DALL-E    | Text-to-image generation     | Text only         | Image  |
| PaLM-E    | Multimodal generation/control| Perceiver + LM    | Text    |



## Limitations of CLIP

- Cannot generate text or images (not generative).
- Sensitive to prompt formulation.
- Struggles with tasks requiring fine-grained reasoning.
- No cross-attention between modalities (unlike Flamingo, etc.).

## Extensions and Successors

- **CLIPCap**: Caption generation using CLIP embeddings.
- **ALIGN (Google)**: Similar method with larger data.
- **BLIP, BLIP-2**: Unified vision-language pretraining and generation.
- **GIT, Flamingo**: Interleaved multimodal Transformers for generation.

## Key Takeaways

- CLIP is a **contrastive, non-generative** model.
- Enables **open-vocabulary, zero-shot vision tasks**.
- Uses **frozen encoders** post-training for many applications.
- Inspires many generative multimodal models by showing the power of large-scale vision-language alignment.

# Flamingo

Interleaved Vision-Language Modeling

## Introduction

- [Flamingo](https://arxiv.org/pdf/2204.14198) is a family of **multimodal few-shot learners** 
- Developed by DeepMind
- Flamingo is a **generative model** that processes interleaved image-text inputs and generates text outputs
   - This differs from CLIP, which performs retrieval or classification in latent space.

In practice:

- Flamingo is built for in-context few-shot reasoning over mixed image-text sequences.
- CLIP is built for similarity scoring ("which text matches this image?"), often used as an encoder or zero-shot classifier.

## Motivation

Prior models either:

- Handled unimodal sequences well (e.g., GPT-3 for text).
- Or required task-specific training to incorporate images (e.g., VL transformers).

**Flamingo’s goal**: Build a **general-purpose multimodal model** that can:

- Take interleaved visual and textual inputs.
- Perform few-shot and zero-shot tasks.
- Work well on open-ended vision-language tasks (e.g., VQA, captioning, dialog).

## Architecture Overview

:::: {.columns}
::: {.column width="50%"}

Flamingo builds on a **frozen pretrained LLM (like Chinchilla)** and adds:

- **Frozen Vision Encoder** (e.g., CLIP ViT).
- **Perceiver Resamplers** to downsample image features into a fixed number of latent tokens.
- **Gated Cross-Attention Layers** inserted between blocks of the language model to fuse image features with text tokens.

### Component Summary:

- 🔷 Frozen: Vision encoder + most LM layers.
- 🟣 Trained: Perceiver resampler + cross-attention modules.

:::
::: {.column width="50%"}

[![](./img/flamingo.png)](https://arxiv.org/pdf/2204.14198#page=4)

:::
::::

## Data Flow in Flamingo

1. **Image Input** → Vision encoder → Perceiver resampler → Latent visual tokens.
2. **Text Input** → Tokenized and embedded.
3. **Interleaving**: Model receives input sequences like:

   ```
   <image> This is a dog. <image> It is playing fetch.
   ```

4. **Fusion**: At cross-attention layers, visual tokens are injected and gated.
5. **Output**: Generated text (caption, answer, etc.)



## Perceiver Resampler

- Converts variable-length visual tokens to a fixed-length set of latent embeddings.
- Uses learnable latent queries and cross-attends over image features.
- Enables scalability and composability across images.



## Cross-Attention for Modality Fusion

- Injected periodically between LLM blocks.
- Gated mechanism allows the model to decide *when* to attend to visual tokens.
- Supports **dynamic multimodal reasoning** (e.g., "look" at the image only when needed).

[![](./img/flamingo-cross-attention.png)](https://arxiv.org/pdf/2204.14198#page=5)



## Few-Shot Prompting

Like GPT-3, Flamingo can do **few-shot learning** via in-context examples:

```
<image> What is in the image?
A dog.
<image> What is in the image?
A cat.
<image> What is in the image?
```

→ Model predicts: "A horse."

No finetuning required — just clever prompting.


## Training Paradigm

- **Frozen backbone**: Vision encoder + LLM.
- **Trainable modules**: Perceiver resampler + cross-attn.
- Trained on large-scale multimodal data (e.g., LAION, CC, HowTo100M).
- **Multitask objective**: generative language modeling conditioned on multimodal context.



## Evaluation Benchmarks

Flamingo achieves SOTA on a wide range of vision-language tasks:

- VQAv2, OKVQA (Visual QA)
- Image captioning (COCO)
- Science QA (text and diagrams)
- Video QA (with frame sequences)



## Comparison with CLIP

| Feature               | CLIP                        | Flamingo                      |
|----------------------|-----------------------------|-------------------------------|
| Modality Fusion      | No (contrastive)            | Yes (cross-attention)         |
| Generation           | No                          | Yes (text generation)         |
| Visual Encoding      | Single image                | Multiple, interleaved images  |
| Training Objective   | Contrastive loss            | Autoregressive LM             |
| Usage                | Retrieval, zero-shot class. | Captioning, VQA, QA, dialog   |


## Strengths

- Interleaved multimodal reasoning.
- Strong few-shot generalization.
- Frozen pretrained components make it modular.
- Minimal tuning — new tasks need only a prompt.



## Limitations

- Large memory footprint.
- Cannot handle non-image modalities (e.g., audio) natively.
- Text-conditioned generation only (cannot generate images).
- Dependency on prompt quality.



## Extensions and Derivatives

- **PaLI, PaLM-E**: Incorporate structured perception + control.
- **GIT**: Combines image encoder + LM with decoder-only modeling.
- **BLIP-2**: Adds pretraining objectives for question answering and captioning.



## Key Takeaways

- Flamingo is a **multimodal decoder** architecture with flexible interleaving.
- Uses **frozen vision + language backbones** for scalability.
- Learns to attend to visual content via gated attention.
- Sets a new standard for **few-shot multimodal learning**.

# PaLI

## What It Is

[**PaLI (Pathways Language and Image)**](https://arxiv.org/pdf/2209.06794) is a Google vision-language model family that reframes many image tasks as one interface: **text generation from image + prompt**.


[![](./img/pali.png)](https://arxiv.org/pdf/2209.06794#page=4)



## Why It Matters

- **Unified task format**: captioning, VQA, OCR, and localization all use prompted text output.
- **Multilingual by design**: supports 100+ languages with cross-lingual transfer.
- **Scales effectively**: larger variants improve performance across benchmarks.


## Core Architecture (Simple View)

- **ViT image encoder** -> visual tokens
- **Text decoder** -> autoregressive output
- **Optional prefix prompt** specifies task and language

Example:

```
<image> caption in Spanish:
```


## Training and Task Framing

- Trained on large multimodal + multilingual corpora.
- Uses standard next-token language modeling.
- Main idea: convert many task types into one generation pipeline.


## PaLI vs Flamingo (Quick Contrast)

- **Flamingo**: interleaved image-text context with cross-attention adapters.
- **PaLI**: single-image + prefix prompt with a unified generation path.
- **Both**: generate text; **PaLI** is especially strong on multilingual scaling.


## Strengths and Limits

**Strengths**

- One model interface for many vision-language tasks
- Strong multilingual capability
- Flexible prompting and structured text outputs

**Limits**

- Large models are compute-heavy
- Performance is prompt-sensitive
- Primarily image-focused (not full temporal multimodal)


## Key Takeaways

- PaLI demonstrates **unified, prompt-based multimodal generation at scale**.
- It is a strong bridge from task-specific systems to general multilingual VLMs.


# LLaVA

## Instruction-Tuned Open Multimodal Assistant

[LLaVA](https://arxiv.org/pdf/2304.08485) is an open recipe for building multimodal assistants that can **see + follow instructions + respond in natural language**.

- **What it does**: image-grounded conversation (VQA, explanation, visual tutoring)
- **Core components**: vision encoder + projector + LLM
- **Training recipe**: visual instruction tuning for assistant-style behavior
- **Why it matters**: practical, reproducible, and widely used in open-source systems

**Use cases**

- UI understanding and workflow assistance
- Document/image question answering
- Educational and accessibility assistants

**Caveat**

- Can hallucinate visual details; grounding checks remain essential


## How LLaVA Works (Data Flow)

1. **Image -> Vision Encoder**
   - A ViT-style encoder (often CLIP ViT) converts pixels into visual embeddings.

2. **Projector -> Language Space**
   - A learned projector maps visual embeddings into tokens the LLM can consume.

3. **Instruction + Visual Tokens -> LLM**
   - The LLM receives both the user prompt and projected visual tokens.

4. **Autoregressive Text Generation**
   - The model generates grounded responses (answers, explanations, step guidance).

Key intuition:

- CLIP-like features provide visual perception.
- Instruction tuning teaches the model how to respond as an assistant.

## Why LLaVA Connects the Whole Lecture

**From taxonomy to practice**

- **CLIP side**: strong visual representation and alignment priors
- **Flamingo side**: multimodal conditioned text generation
- **LLaVA contribution**: instruction-tuned behavior for real assistant workflows

**Agentic relevance**

- Enables observe -> reason -> respond loops over images and interfaces
- Useful in UI agents, document assistants, and multimodal copilots

**Important limitation**

- Good fluency does not guarantee faithful grounding
- Add retrieval checks, tool verification, or human oversight for high-stakes tasks
