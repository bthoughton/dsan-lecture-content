# Lecture: PaLI – Pathways Language and Image

## Introduction

**PaLI** (Pathways Language and Image) is a family of large-scale multimodal models developed by Google Research. It generalizes image-to-text tasks through **multilingual, multimodal pretraining**, unifying vision and language under a single generative framework.

---

## Motivation

Previous models (e.g., Flamingo, SimVLM) were powerful but often limited in:

- Language coverage (usually English-only),
- Task-specific design,
- Limited scalability across both vision and multilingual text.

**PaLI’s goal**: Create a **universal vision-language model** that can:

- Understand and generate **multilingual text**,
- Handle various image-to-text tasks (captioning, VQA, OCR, etc.),
- Scale efficiently to hundreds of billions of parameters.

---

## Architecture Overview

PaLI is built on top of Google's **Pathways** infrastructure and uses:

- A **Vision Transformer (ViT)** as the image encoder.
- A **Text Transformer Decoder** that generates multilingual output.
- Optional **Prefix Language Tokens** to specify task type or language.

### Component Summary:

- 🟩 Vision encoder: Processes image into patch tokens.
- 🟦 Text decoder: Generates output conditioned on image + text.
- 🧩 Unified decoder-only model for all tasks.

---

## Data Flow in PaLI

1. **Image Input** → ViT → Visual patch embeddings.
2. **Optional Task Prefix** (e.g., “caption in French:”).
3. **Decoder** generates output autoregressively:

   ```
   Input: <image> caption in Spanish:
   Output: Un perro jugando con una pelota.
   ```

Tasks are cast as **text generation** conditioned on the image and an optional prefix.

---

## Training Paradigm

- Pretrained on **massive multimodal data**, including:
  - Web image-text pairs (e.g., CC12M, WIT),
  - OCR-labeled documents,
  - Multilingual corpora (over 100 languages).
- Objective: **Standard language modeling** — predict next token given previous tokens and image.
- No special contrastive or ranking loss.

---

## Tasks Unified under Text Generation

All tasks share the same format:

- 🔤 **Image captioning**: `<image> caption: ...`
- ❓ **VQA**: `<image> Q: What is the color of the car? A: ...`
- 📖 **OCR / Reading**: `<image> read text: ...`
- 🗺️ **Object localization**: Generate bounding boxes as text tokens.

This makes PaLI highly flexible and easy to prompt.

---

## Multilingual Modeling

- PaLI handles **over 100 languages**, using a shared tokenizer.
- Unified model enables **cross-lingual transfer**.
- Supports **code-switching**, image-question in one language, answer in another.

---

## Scaling and Model Variants

Several model sizes are trained (e.g., PaLI-3B, PaLI-17B, PaLI-55B):

- Larger models improve performance across tasks.
- PaLI-17B and 55B match or exceed SOTA on many vision-language benchmarks.

---

## Evaluation Benchmarks

PaLI is evaluated on a wide range of tasks:

- **Captioning**: COCO, NoCaps
- **VQA**: VQAv2, TextVQA, GQA
- **OCR**: IAM, DocVQA, CORD
- **Cross-lingual**: xGQA, MUCOK, MaRVL
- **Image classification**: zero-shot transfer to ImageNet, etc.

---

## Comparison with Flamingo

| Feature                | Flamingo                     | PaLI                         |
|------------------------|------------------------------|------------------------------|
| Input Modality         | Interleaved images + text    | Single image + text prefix   |
| Language Support       | Primarily English            | 100+ languages               |
| Output Type            | Text generation              | Text generation              |
| Architecture           | Frozen LM + adapters         | Unified ViT → Decoder        |
| Task Format            | Prompted                     | Prompted                     |

---

## Strengths

- Unified, scalable architecture across many tasks.
- Strong **multilingual capabilities**.
- No need to fine-tune for different vision-language tasks.
- Flexibly outputs structured answers (bounding boxes, OCR, etc.).

---

## Limitations

- Focused on image input only (no video/audio).
- Decoder-only architecture can be slow at inference time.
- Requires substantial compute for large-scale pretraining.
- Prefix prompting can be brittle without task engineering.

---

## Extensions and Derivatives

- **PaLI-X**: Further scaling with sparse expert models.
- **UDOP**: Unified decoder for documents with layout modeling.
- **PaLM-E**: Incorporates embodied control and robotics.
- **Gemini** (2023): Google's next-gen multimodal foundation model — influenced by PaLI.

---

## Key Takeaways

- PaLI is a **universal image-to-text model** supporting multilingual tasks.
- Scales well to large model sizes and diverse inputs.
- Casts all tasks as **text generation**, simplifying training and inference.
- Highlights the power of **prompt-based multimodal learning** at scale.
