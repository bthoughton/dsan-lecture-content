## A story: the tests were green

Production **document extraction**: a computer vision workflow. I ask an agent
to test it.

It writes tests. **The tests pass.**

. . .

**What's wrong?**

. . .

The agent was **simulating the response from the VLM.**

The tests were green and **nothing was tested.**

. . .

This happens *routinely*. And notice the shape of the failure: the agent didn't
write broken code; it wrote code that **reported success without doing the
work.** The output was indistinguishable from real success.

::: {.callout-important}
## You will meet this again in this course
A model scoring 0.98 because the split leaked. A metric that looks great because
the classes are imbalanced. A loss that optimized something *adjacent* to what
you cared about.

These are all the same kind of failure.

# AI, ML, and deep learning are not synonyms

- These three words get used interchangeably in the press
- They are **nested**, not equivalent
- Knowing which one you mean is the first sign you know what you're talking about

## Artificial intelligence

**AI:** any effort to automate intellectual tasks normally performed by humans.

This is an extremely broad category. It includes all forms of inorganic
"intelligence"; by this definition, an old pocket calculator technically
qualifies.

![](images/2023-01-09-18-22-49.png){width=550 .dh400}

<sup>Source: Chollet</sup>

## Traditional programming

Based on **explicit instructions**: the human tries to tell the machine what to
do for *every* possible situation. In practice that means hard-coded rules, a
large pile of if-statements. This is **symbolic AI**, roughly 1950s–1980s.

![](images/fig-classical-programming.png){width=700 .dw70}

This does not scale to large, messy problems. Try writing the rules for "is
there a cat in this photo."

**So: can the algorithm learn from data instead of being told what to do?**

## Machine learning

**ML:** the machine learns the rules from data, given examples of what's expected.

![](images/fig-machine-learning.png){width=700 .dw70}

- **Definition 1**: searching for useful *representations* of some input data,
  within a predefined space of possibilities, using guidance from a feedback
  signal. *(Chollet)*
- **Definition 2**: methods that leverage data to improve performance.

**Training** is the automatic search for better representations.

## The machine learning paradigms {.smaller}

![](images/2023-01-09-18-11-03.png){width=800 .dh320}

- **Supervised**: you have inputs *and* the right answers. *(This is us, mostly.)*
- **Semi-supervised**: a few labels plus a large amount of unlabeled data.
  This is closer to how most modern systems are actually built: pretrain on
  everything, then fine-tune on the labels you have.
- **Unsupervised**: inputs only; find structure. Objectives are fuzzier and
  success is harder to measure.
- **Reinforcement learning**: an agent takes actions in an environment and
  learns from reward. *(DSAN-6650.)*

## Supervised learning: classification vs regression {.smaller}

The target variable decides which one you're doing.

**Classification**: the target is a category

![](images/2023-01-16-17-13-25.png){width=520 .dw55}

**Regression**: the target is a continuous number

![](images/2023-01-16-17-14-16.png){width=520 .dw52}

## Deep learning

**Deep learning:** learning **successive layers of increasingly meaningful
representations.**

![](images/fig-representation-chain.png){width=680 .dw85}

**Depth** is the number of layers, which is the number of successive
representations. Shallow learning uses one or two; deep learning uses more.

Two sentences worth writing down:

- **Shallow ML** has only *one* transformed representation: linear regression,
  logistic regression.
- **Deep learning is a form of parametric statistical learning.** A neural
  network is a parameterized function fit to data, *exactly like* linear or
  logistic regression.

That second sentence is the whole reason the next section looks the way it does.

# Six components, every model

Every model in this course, from today's linear regression to Week 8's
transformer, is the **same six pieces**:

1. **Data**: features and a target
2. **A parametric model**: a function with adjustable knobs
3. **Residuals**: how wrong it is, per example
4. **A loss function**: the rule that turns that wrongness into a score
5. **Splits**: train / validation / test, so you don't fool yourself
6. **An optimizer**: the thing that turns the knobs

Learn them once and the rest of the semester is variations. This is also the
vocabulary you use to **direct an agent** and **audit** what it hands back.

<sup>Two more join the list in Week 4: **regularization** and **hyper-parameter
tuning**. Six is enough to get you a working model.</sup>

## Component 1: Data

**Supervised** data gives you a mapping from features to a target:

$$\mathbf{x} \;(\text{features}) \;\longrightarrow\; y \;(\text{target})$$

- $y$ is **discrete** → classification
- $y$ is **continuous** → regression

**The mapping notation**: you'll see this all semester:

$$\underbrace{\mathbb{R}^{2}}_{\text{input layer}} \longrightarrow \underbrace{\mathbb{R}^{1}}_{\text{output layer}}$$

Two features in, one number out. The shape of the data *is* the shape of the
problem.

## Component 2: A parametric model {.smaller}

A **parametric model** is a function with adjustable knobs:

$$y_{\text{pred}} = M(\mathbf{x} \mid \mathbf{w})$$

- $\mathbf{x}$: the input (given to you)
- $\mathbf{w}$: the **parameters** / weights (what training changes)

![](images/2023-01-31-11-26-02.png){width=600 .dh320}

Training means: **find the $\mathbf{w}$ that makes the predictions good.**

## Example: linear regression

The model you already know, written in this vocabulary.

**Univariate**: 2 parameters

$$y = w_0 + w_1 x$$

**Multivariate**: $n+1$ parameters

$$y = w_0 + w_1 x_1 + w_2 x_2 + \dots + w_n x_n$$

**Vector output**: a matrix of parameters

$$\mathbf{y} = \mathbf{W}\mathbf{x} + \mathbf{b}$$

Hold onto this. We come back to it at the end of today.

## Parameters = flexibility

More parameters means the model can bend into more shapes.

![](images/2023-01-24-09-12-22.png){width=680 .dw51}

That is **not** automatically good. A model flexible enough to fit the noise will
fit the noise, and then fail on data it hasn't seen. That's **overfitting**, and
we'll make it concrete in a few slides.

## Component 3: Residuals

A **residual** is how wrong the model is on one example:

$$r_i = y_{\text{pred},i} - y_{\text{target},i}$$

![](images/2023-01-31-11-43-11.png){width=700 .dw58}

Residuals are per-example. Next we need to turn *all* of them into a **single
number** we can push downhill.

## Component 4: Loss {.smaller}

A **loss** scores a *single* example. The **cost** collapses every example's loss
into the one scalar that scores a whole parameter choice:

$$\ell_i = \ell\big(y_i, \hat{y}_i\big)
\qquad\qquad
\mathcal{L}(\mathbf{w}) = \frac{1}{N}\sum_{i=1}^{N} \ell_i$$

The two words are used interchangeably in practice, PyTorch included. Keep them
straight when you are reading a formula and deciding what is averaged over what.

The loss you pick is determined by the **problem type**:

| Problem type | Loss | In PyTorch (Week 3) |
|---|---|---|
| Regression | Mean squared error | `nn.MSELoss` |
| Regression, robust to outliers | Mean absolute error | `nn.L1Loss` |
| Binary classification | Binary cross-entropy | `nn.BCEWithLogitsLoss` |
| Multi-class classification | Categorical cross-entropy | `nn.CrossEntropyLoss` |

Know those four. There are dozens more, with new ones every year, so the
durable skill is knowing how to **find and choose** a loss that matches your
problem, and noticing when something has chosen the wrong one for you.

## Metrics are not losses

- **Loss**: what the optimizer minimizes. Must be differentiable.
- **Metric**: what *you* actually care about, reported to humans. Accuracy, F1,
  precision at 90% recall, dollars saved.

These are frequently **not the same function**. Accuracy isn't differentiable, so
you can't train on it; you train on cross-entropy and *report* accuracy.

::: {.callout-important}
## This gap is where things go wrong
Choosing the loss is **your** job. An agent will optimize whatever you gave it,
forever, without ever mentioning that it doesn't match the thing you said you
cared about.
:::

## Regression loss: MSE and MAE

::: {.content-hidden when-format="revealjs"}

**Mean squared error**: the default

$$\text{MSE} = \frac{1}{N}\sum_{i=1}^{N}\left(y_i - \hat{y}_i\right)^2$$

Squaring punishes big misses hard, and makes the loss smooth and easy to
differentiate.

**Mean absolute error**: more forgiving of outliers

$$\text{MAE} = \frac{1}{N}\sum_{i=1}^{N}\left|y_i - \hat{y}_i\right|$$

![](images/2024-09-05-13-59-25.png){width=650}

:::

::: {.content-visible when-format="revealjs"}
![](images/2024-09-05-13-59-25.png){.dh500}

**MSE** squares the errors, so a few big misses dominate, and the loss stays
smooth and easy to differentiate. **MAE** does not, so it is far more forgiving
of outliers.
:::

## Classification loss: binary cross-entropy

For binary classification the model outputs a **probability** $\hat{y} \in (0,1)$:

$$\text{BCE} = -\frac{1}{N}\sum_{i=1}^{N}\Big[\, y_i \log \hat{y}_i + (1-y_i)\log(1-\hat{y}_i) \,\Big]$$

![](images/fig-bce-penalty.png){width=660 .dw45}

**The intuition worth keeping:** the $\log$ blows up as you approach a confident
wrong answer. Being *confidently* wrong is punished far more than being *unsure*
and wrong, which is exactly what you want from something making decisions.

## The loss surface

$\mathcal{L}(\mathbf{w})$ is a function of the parameters. Plot it and you get a **surface**.

![](images/2023-01-31-11-57-42.png){width=560 .dh400}

**Training is finding the bottom of this surface.**

Every optimizer in this course is a different strategy for walking downhill.

## Component 5: Splits

You cannot evaluate a model on the data it learned from. It has already seen the
answers.

![](images/2023-02-07-12-24-02.png){width=760 .dh320}

- **Train**: the model learns from this
- **Validation**: you tune your choices against this
- **Test**: touched **once**, at the very end, to report a number

## Why the test set stays untouched

Every time you look at the test set and then change something, you leak a little
information into your model. Do it enough and the test score stops being an
honest estimate of how the model performs on data it has never seen.

- Tune on **validation**. Report on **test**. Once.
- *k*-fold cross-validation exists and is excellent, but deep learning retraining
  is expensive, so a single held-out split is the norm here.

::: {.callout-warning}
## The most common way this breaks
You split **after** doing something that mixes the sets: normalizing over the
full dataset, augmenting before splitting, or splitting random frames out of the
same video. The score goes up and means nothing.
:::

## Overfitting, concretely {.smaller}

This is the single most important plot in machine learning.

![](images/fig-overfitting.png){width=560 .dh380}

- **Training loss** keeps going down: the model keeps memorizing
- **Validation loss** bottoms out, then **turns back up**
- That turning point is where the model stopped learning the pattern and started
  learning the noise

**Early stopping** is the simplest fix: stop at the bottom.

::: {.content-hidden when-format="revealjs"}
::: {.callout-warning}
Reading this plot correctly is a skill **no agent has.** It will report the
training loss and tell you things are going great.
:::
:::

## Component 6: Optimization

We need the $\mathbf{w}$ that minimizes $\mathcal{L}(\mathbf{w})$. There are two ways.

**Option 1: analytically.** Pen and paper. For linear regression there's a
closed-form answer:

$$\hat{\beta} = \left(\mathbf{X}^{\top}\mathbf{X}\right)^{-1}\mathbf{X}^{\top}\mathbf{y}$$

**Option 2: numerically.** Start somewhere, repeatedly step downhill.

::: {.callout-note}
## Why gradient descent exists
For a neural network there **is no closed form.** Numerical optimization isn't a
stylistic preference; it's the only option. That's the entire reason we care
about gradients.
:::

## Gradients: only as much as we need {.smaller}

This is not a calculus course, and **this slide is the whole calculus requirement.**

- A **partial derivative** $\dfrac{\partial \mathcal{L}}{\partial w_i}$ is the slope of the
  loss if you nudge *one* parameter and hold the rest still.
- The **gradient** is just the vector of all of them:

::: {.content-hidden when-format="revealjs"}
$$\nabla \mathcal{L}(\mathbf{w}) = \left(\frac{\partial \mathcal{L}}{\partial w_0}, \frac{\partial \mathcal{L}}{\partial w_1}, \dots, \frac{\partial \mathcal{L}}{\partial w_N}\right)$$
:::

![](images/2023-01-26-09-25-50.png){width=700 .dw58}

- It points in the direction of **steepest increase**. So $-\nabla \mathcal{L}$
  points downhill.
- At the bottom of a valley, the surface is flat: $\nabla \mathcal{L}(\mathbf{w}) = 0$.

That's all you need.

## Gradient descent

::::: {.content-hidden when-format="revealjs"}

![](images/GD.gif){width=560}

Repeat until converged:

$$\mathbf{w}_{\text{new}} = \mathbf{w}_{\text{old}} - \underbrace{\lambda}_{\text{learning rate}} \nabla \mathcal{L}(\mathbf{w}_{\text{old}})$$

1. Look at the slope where you're standing
2. Take a step downhill
3. Repeat

The **learning rate** $\lambda$ is how big a step you take. Too small and you
never arrive; too big and you bounce out of the valley. **Week 4 is entirely
about doing this well.**

## All six components, in one cell {.smaller .codetight}

Not walking through this line by line; it is in the handout too. The one line
worth finding is inside `loss`, the only place the choice of MSE appears.

```python
#| echo: true
import numpy as np
rng = np.random.default_rng(6600)

# ---- Component 1: DATA (noisy measurements of a decaying oscillation)
x = np.linspace(0, 10, 120)
A_true, k_true = 3.0, 0.35
y = A_true * np.exp(-k_true * x) * np.sin(2.0 * x) + rng.normal(0, 0.18, x.size)

# ---- Component 2: MODEL (two parameters, so the cost surface is plottable)
def model(x, A, k):
    return A * np.exp(-k * x) * np.sin(2.0 * x)

# ---- Components 3 & 4: RESIDUALS -> ONE SCALAR (mean over every example)
def loss(A, k):
    return np.mean((y - model(x, A, k)) ** 2)      # <- this is MSE

# ---- Component 6: OPTIMIZER (plain gradient descent, numeric gradient)
def grad(A, k, h=1e-5):
    return ((loss(A + h, k) - loss(A - h, k)) / (2 * h),
            (loss(A, k + h) - loss(A, k - h)) / (2 * h))

A, k, lr = 0.35, 1.05, 0.06        # a deliberately bad starting guess
path = [(A, k, loss(A, k))]
for _ in range(1500):
    gA, gk = grad(A, k)
    A -= lr * gA
    k -= lr * gk
    path.append((A, k, loss(A, k)))
path = np.array(path)

print(f"start loss   {path[0, 2]:.4f}")
print(f"final loss   {path[-1, 2]:.4f}")
print(f"recovered    A={path[-1, 0]:.2f}, k={path[-1, 1]:.2f}   (true: A={A_true}, k={k_true})")
```

Walk it back through the six components:

| **2. Model** | $A e^{-kx}\sin(2x)$: two parameters |
| **3. Residuals** | $y - \hat{y}$ at every point |
| **4. Loss** | squared error per point, averaged into one number |
| **5. Splits** | *(skipped here; we'll need it the moment the model is flexible)* |

## The punchline
**Nothing structural changes when the model becomes a neural network.** The model
function just gets bigger: thousands or billions of parameters instead of two.

The loss surface stops being something you can draw. The six components stay
exactly the same.

## So what makes it a neural network?

Two moves. That's the entire idea.

:::: {.columns}
::: {.column width="50%"}
### 1. Stack them
Take many of these units and arrange them in **layers**, feeding left to right.

![](images/2023-03-21-10-46-50.png){width=420 .dw100}
:::
::: {.column width="50%"}
### 2. Put a nonlinearity between the layers
Without it, stacked linear layers collapse back into **one** linear layer and you
have gained nothing.

![](images/2023-03-21-10-43-20.png){width=380 .dw89}
:::
::::

**Stack, and add a nonlinearity.** Everything else this semester is a clever
variation on those two moves: CNNs, transformers, diffusion models.

## Why stacking helps

Neural networks are **universal function approximators**: with enough units, they
can approximate essentially any continuous function.

![](images/2023-03-20-22-26-18.png){width=680 .dh400}

That is why *one* family of models covers the *entire* problem taxonomy from
earlier: scalar regression, vector regression, binary and multi-class
classification, generation.
